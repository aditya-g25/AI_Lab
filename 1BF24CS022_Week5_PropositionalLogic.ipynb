```python
from itertools import product

def implies(a, b):
    return not a or b

def check_kb(P, Q, R):
    return implies(Q, P) and implies(P, not Q) and (Q or R)

def check_query(query, P, Q, R):
    return query(P, Q, R)

print("P Q R | Q->P P->~Q Q|R | KB | R | R->P | Q->R")
print("-" * 52)

models = []

for P, Q, R in product([False, True], repeat=3):
    s1 = implies(Q, P)
    s2 = implies(P, not Q)
    s3 = Q or R

    KB = s1 and s2 and s3
    query1 = R
    query2 = implies(R, P)
    query3 = implies(Q, R)

    print(
        int(P), int(Q), int(R), "|",
        int(s1), "   ", int(s2), "   ", int(s3), "|",
        int(KB), " |", int(query1), " |",
        int(query2), "   |", int(query3)
    )

    if KB:
        models.append((P, Q, R))

print("\nModels where KB is true:")
for P, Q, R in models:
    print(f"P={int(P)}, Q={int(Q)}, R={int(R)}")

print("\nKB entails R:", all(R for P, Q, R in models))
print("KB entails R -> P:",
      all(implies(R, P) for P, Q, R in models))
print("KB entails Q -> R:",
      all(implies(Q, R) for P, Q, R in models))
```
