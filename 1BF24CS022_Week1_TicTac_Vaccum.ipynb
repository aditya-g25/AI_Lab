```python
import random

A = random.randint(0, 1)
B = random.randint(0, 1)

print("Initial State:")
print("A =", A)
print("B =", B)

vacuum = random.randint(0, 1)

score = 0

if vacuum == 0:
    print("\nVacuum is at Location A")

    if A == 1:
        print("A is Dirty -> Suck")
        A = 0
        score += 1
    else:
        print("A is Clean")

    print("Moving to Location B")

    if B == 1:
        print("B is Dirty -> Suck")
        B = 0
        score += 1
    else:
        print("B is Clean")

else:
    print("\nVacuum is at Location B")

    if B == 1:
        print("B is Dirty -> Suck")
        B = 0
        score += 1
    else:
        print("B is Clean")

    print("Moving to Location A")

    if A == 1:
        print("A is Dirty -> Suck")
        A = 0
        score += 1
    else:
        print("A is Clean")

print("\nFinal State:")
print("A =", A)
print("B =", B)

print("Performance Measurement:", score)
```

```python
board = [[" " for _ in range(3)] for _ in range(3)]

current_player = "X"

def display_board():
    print("\n-------------")
    for row in board:
        print("| " + " | ".join(row) + " |")
        print("-------------")

def check_winner(player):
    for i in range(3):
        if (board[i][0] == player and
            board[i][1] == player and
            board[i][2] == player):
            return True

    for i in range(3):
        if (board[0][i] == player and
            board[1][i] == player and
            board[2][i] == player):
            return True

    if (board[0][0] == player and
        board[1][1] == player and
        board[2][2] == player):
        return True

    if (board[0][2] == player and
        board[1][1] == player and
        board[2][0] == player):
        return True

    return False

for turn in range(9):

    display_board()

    print("\nPlayer", current_player)

    while True:
        row = int(input("Enter row (1-3): ")) - 1
        col = int(input("Enter column (1-3): ")) - 1

        if row < 0 or row > 2 or col < 0 or col > 2:
            print("Invalid position. Enter values from 1 to 3.")

        elif board[row][col] != " ":
            print("Position already occupied. Try again.")

        else:
            break

    board[row][col] = current_player

    if check_winner(current_player):
        display_board()
        print("\nPlayer", current_player, "wins!")
        break

    if current_player == "X":
        current_player = "O"
    else:
        current_player = "X"

else:
    display_board()
    print("\nGame Draw!")
```
