```python
def dls(state, goal, depth, path):
    if state == goal:
        return True

    if depth == 0:
        return False

    zero = state.index(0)
    row = zero // 3
    col = zero % 3

    moves = []

    if row > 0:
        moves.append(zero - 3)
    if row < 2:
        moves.append(zero + 3)
    if col > 0:
        moves.append(zero - 1)
    if col < 2:
        moves.append(zero + 1)

    for new_zero in moves:
        new_state = state[:]
        new_state[zero], new_state[new_zero] = \
            new_state[new_zero], new_state[zero]

        if new_state not in path:
            path.append(new_state)

            if dls(new_state, goal, depth - 1, path):
                return True

            path.pop()

    return False


print("Enter Initial State:")
initial = list(map(int, input().split()))

print("Enter Goal State:")
goal = list(map(int, input().split()))

depth_limit = int(input("Enter Depth Limit: "))

path = [initial]

if dls(initial, goal, depth_limit, path):
    print("\nSolution Found!\n")

    for state in path:
        for i in range(0, 9, 3):
            print(state[i:i+3])
        print()
else:
    print("\nSolution not found within the given depth limit.")
```
