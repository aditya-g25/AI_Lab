```python
def calculate_cost(board):
    cost = 0
    n = len(board)

    for i in range(n):
        for j in range(i + 1, n):
            if (board[i] == board[j] or
                abs(board[i] - board[j]) == abs(i - j)):
                cost += 1

    return cost


def hill_climbing(board):
    n = len(board)
    iteration = 0

    while True:
        current_cost = calculate_cost(board)

        print("\n" + "=" * 45)
        print("Iteration:", iteration)
        print("Current state:", board)
        print("Current cost:", current_cost)

        if current_cost == 0:
            print("\nGOAL STATE REACHED!")
            print("Final board:", board)
            print("Final cost:", current_cost)
            print_board(board)
            return
        best_board = None
        best_cost = current_cost
        best_pair = None

        print("\nNeighbor states and costs:")
        for i in range(n):
            for j in range(i + 1, n):
                neighbor = board.copy()
                neighbor[i], neighbor[j] = (
                    neighbor[j], neighbor[i]
                )

                neighbor_cost = calculate_cost(neighbor)

                print(
                    f"Swap (x{i}, x{j}) -> {neighbor}"
                    f" | Cost = {neighbor_cost}"
                )
                if neighbor_cost < best_cost:
                    best_cost = neighbor_cost
                    best_board = neighbor
                    best_pair = (i, j)
        if best_board is None:
            print("\nLocal minimum reached!")
            print("No neighbor has a lower cost.")
            print("Algorithm terminates without a solution.")
            return

        print("\nSelected swap:", best_pair)
        print("Next state:", best_board)
        print("Next cost:", best_cost)

        board = best_board
        iteration += 1


def print_board(board):
    print("\nChessboard:")
    n = len(board)

    for row in range(n):
        for col in range(n):
            if board[col] == row:
                print(" Q ", end="")
            else:
                print(" . ", end="")
        print()

print("HILL CLIMBING - 4 QUEENS")
print("Enter row positions from 0 to 3.")

values = input("Enter 4 row positions").split()

if len(values) != 4:
    print("Error: Enter exactly 4 values.")
else:
    try:
        board = [int(x) for x in values]

        if sorted(board) != [0, 1, 2, 3]:
            print("Error: Enter each row number 0, 1, 2, 3 once.")
        else:
            hill_climbing(board)

    except ValueError:
        print("Error: Please enter integer values only.")


```
